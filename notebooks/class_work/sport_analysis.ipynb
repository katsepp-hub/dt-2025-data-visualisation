```python
import pandas as pd

df= pd.read_csv("sports.csv")
df.head() # this is to show the first rows
```

```python
df.describe() # generate some statistics
```

Box plot for each athlete to find the number of years of practice

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.ylabel("Years")
plt.title("Years of practice per athlete")
plt.show
```

Age of each athlete

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.ylabel("Age")
plt.title("Age of athlete")
plt.show
```

Compare Years of practice with Age

plt.scatte(x,y) makes points for each pair of values
"s" controls the size of each bubble
s=df["Years_of_practice"]*30 means: if a studemt has more years of practice, their bubble is larger. *30 is a scaling factor to make the bubble big.
alpha=0.6 make the bubbles 60% transparent.

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Years")
plt.title("Age vs years of practice")
plt.show
```

Average age per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("Average age per sport")
plt.ylabel("Average age")
plt.xlabel("Sport")
plt.show
```
